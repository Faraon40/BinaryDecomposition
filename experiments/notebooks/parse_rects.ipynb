```python
# --- 🟩 Krok 1: Import knižníc ---
import pandas as pd
import re

# --- 🟩 Krok 2: Načítanie dát ---
# Súbor "research_results.txt" musí byť v rovnakom priečinku ako tento notebook
input_file = "data/research_results.txt"

data = []

with open(input_file, "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue

        # Oddelíme názov obrázka (prvá časť pred prvou medzerou)
        parts = line.split(maxsplit=1)
        name = parts[0]

        # Extrahujeme všetky číselné časti
        nums = re.findall(r"\d+", parts[1])

        # --- 🟦 Krok 3: Rekonštrukcia čísel ---
        # Spojí "1" + "792" -> 1792, ale nechá 3-ciferné ako 956
        joined = []
        i = 0
        while i < len(nums):
            if len(nums[i]) == 1 and i + 1 < len(nums):
                joined.append(int(nums[i] + nums[i+1]))
                i += 2
            else:
                joined.append(int(nums[i]))
                i += 1

        # --- 🟨 Krok 4: Korekcia dĺžky riadku ---
        # Ak chýbajú čísla, doplní None, ak je ich viac, oreže
        while len(joined) < 4:
            joined.append(None)
        joined = joined[:4]

        # Uložíme riadok
        data.append([name] + joined)

# --- 🟩 Krok 5: Tvorba DataFrame a export ---
df = pd.DataFrame(data, columns=["image_title", "GDM", "M-GDM", "SBD", "CBD"])

# Uloženie do CSV
output_file = "rectangles_fixed.csv"
df.to_csv(output_file, index=False)

print("✅ Hotovo! Dáta boli uložené do:", output_file)
print("\nUkážka prvých riadkov:\n")
display(df.head(10))
```
